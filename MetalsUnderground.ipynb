**Research Question and Hypotheses**

1. Do soil pollutant concentrations (P2O5, PbO, SO3, ZnO ) decrease with increasing distance from the road?


   Hypothesis: Pollutant concentrations are highest near the road and decline with distance, consistent with roadside deposition.


2. How do soil quality indicators vary with distance from the road?

   Hypothesis: Soil farther from the road has lower pollutant concentrations and higher organic matter and nitrogen contents.


**Sampling Design**

Three spatially separated transects will be established perpendicular to a comparable stretch of road
located within an area large enough to be adequately represented by roughly 21 samples in total. Each transect will
include seven locally appropriate and accessible sampling distances. Sampling depth will remain
consistent at 5 cm. Vegetation, recent disturbance, location, soil moisture, and temperature will be
recorded

**Laboratory Analyses**

Samples will be prepared according to laboratory protocols. X-ray fluorescence (XRF) will quantify
P2O5, PbO (air pollution), SO3, ZnO (tire rubber particulates), Ca (concrete/asphalt) as well as other
pollutants. Soil pH will provide context for metal retention and potential mobility. Elemental analysis
(EA) will determine total nitrogen and carbon; total nitrogen will be treated as a nutrient-status indicator
rather than a direct measure of nutrient availability. Additionally, SoliTOC could be useful as it
determines exact TOC (Total Organic Carbon) and TIC (Total Inorganic Carbon), and the ROC
(Residual Oxidized Carbon) pool (based on older bitumen rubber)

Sieving and milling of samples and drying at 60°C, at temperatures above teh decomposition of OM can start. Freez-Drying will take too long

**Data Analysis and Scope**

Metal concentrations and soil quality indicators will be plotted against distance. Analyses will examine distance trends and relationships between metals and pH, accounting for transect grouping. The study will assess soil quality indicators; demonstrating changes in productivity would require an additional measure, such as soil respiration
